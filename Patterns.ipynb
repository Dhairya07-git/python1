```python
# Disarium number: 175: 1^1+7^2+5^3=175
n=input("Enter number: ")
count=len(n)
temp=int(n)
dis=temp
sum=0
while temp>0:
    dig=temp%10
    sum+=(dig**count)
    count-=1
    temp//=10
if(sum==dis):
    print(dis,"is a disarium number")
else:
    print(dis,"is not a disarium number")
```

```python
n=int(input("Enter number: "))
for i in range(n):
    for j in range (i+1):
        print("*",end=" ")
    print()
```

```python
n=int(input("Enter number: "))
for i in range(n):
    for j in range(i+1):
        if ((i+j)%2==0):
            print(1,end=" ")
        else:
            print(0,end=" ")
    print()
```
