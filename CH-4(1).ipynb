- ## Important Functions of String
  1) len() - gives length of string
  2) lstrip() - removes spaces from left side
  3) rstrip() - removes spaces from right side
  4) strip() - removes spaces from all side
  5) lower() - converts whole string to lowercase
  6) upper() - converts whole string to uppercase
  7) swapcase() - converts all lower to upper and all upper to lower
  8) title() - turns first alphabet of every word to capital
  9) capitalize() - turns first alphabet of every line to capital
- ## Check Functions
  1) isalnum() - gives true if string has character or number (gives false if there is space)
  2) isalpha() - gives true if string has character
  3) isnumeric() - gives true in digits and fractions
  4) isdigit() - gives true in digits but not in
  5) islower() - checks if string is in lowercase
  6) isupper() - checks if string is in uppercase
  7) istitle() - checks if every word has 1st alphabet capital
  8) isidentifier()
  9) isspace()

### Write a python program to count all the letters, digits and special symbols from a given string

```python
s="Py $@th%$13on4"
a,b,c,low,up,sp=0,0,0,0,0,0
for i in s:
    if i.isalpha():
        a+=1
        if i.isupper():
            up+=1
        else:
            low+=1
    elif i.isdigit():
        b+=1
    elif i==" ":
        sp+=1
    else: 
        c+=1
print("Number of alphabets:",a)
print("Number of alphabets in lowercase:",low)
print("Number of alphabets in uppercase:",up)
print("Number of spaces:",sp)
print("Number of digits:",b)
print("Number of special characters:",c)

sum=0
for i in s:
    if i.isdigit():
        sum+=int(i)
print("average of digits in the string:",sum/b)
```
