- ## Important Functions of String

  2) lstrip() - removes spaces from left side
  3) rstrip() - removes spaces from right side
  4) strip() - removes spaces from all side
  5) lower() - converts whole string to lowercase
  6) upper() - converts whole string to uppercase
  7) swapcase() - converts all lower to upper and all upper to lower
  8) title() - turns first alphabet of every word to capital
  9) capitalize() - turns first alphabet of every line to capital

  4) isdigit() - gives true in digits but not in
  5) islower() - checks if string is in lowercase

  9) isspace()

### Write a python program to count all the letters, digits and special symbols from a given string

```python
s="Py $@th%$13on4"
a,b,c,low,up,sp=0,0,0,0,0,0
for i in s:
    if i.isalpha():
        a+=1
        if i.isupper():
            up+=1
        else:
            low+=1
    elif i.isdigit():
        b+=1
    elif i==" ":
        sp+=1
    else: 
        c+=1
print("Number of alphabets:",a)
print("Number of alphabets in lowercase:",low)
print("Number of alphabets in uppercase:",up)
print("Number of spaces:",sp)
print("Number of digits:",b)
print("Number of special characters:",c)

sum=0
for i in s:
    if i.isdigit():
        sum+=int(i)
print("average of digits in the string:",sum/b)
```

- Find() -> Returns the index of first occurence of the given substring if it is not available then we will get -1

```python
s="Learning Python is Very Easy"
print(s.find("Python"))
print(s.find("python"))
print(s.find("i"))
print(s.find("h",2,15)) #It finds between index 2 and 15
print(s.find("h",2,10))
```

- Count() -> Returns the number of occurences of the given substring in the string and returns 0 if couldn't find

```python
s="acvsdbwhedazbns,alkhf"
print(s.find(","))
print(s.count("a"))
print(s.count("a",0,8))
```

- Translate() with maketrans function 

```python
import string
s="Py$t@hOn is very ea$sy"
print(string.punctuation)
x1= s.maketrans("","",string.punctuation)
x2= s.maketrans(" "," ",string.punctuation)
print(x1)
print(x2)
print(len(x1))
print(len(x2))
print(s.translate(x1))
print(s.translate(x2))
x3= s.maketrans("","","@")
print(x3)
print(s.translate(x3))
x4= s.maketrans("","","$@ ")
print(x4)
print(s.translate(x4))
print(s.translate(s.maketrans("s","p"))) #It replaces s with p
print(s.translate(s.maketrans("asnO","epmo"))) #It replaces a with e, s with p, n with m and O with o 
print(s.translate(s.maketrans("sa","p"))) #length of both parameters must be equal. If not then it gives error
```

- ### Write a program to merge the characters of 2 string into a single string by taking character's alter value

```python
s1=input("Enter string 1:")
s2=input("Enter string 2:")
s=""
i,j=0,0
while i<len(s1) or j<len(s2):
    s=s+s1[i]
    i+=1
    s=s+s2[j]
    j+=1
print(s)
# Both the strings should be equal then only it works
```
