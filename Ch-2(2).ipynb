```python
# Write a program to check if given number is a happy number
""" A happy number is a positive integer that eventually reaches 1 when you repeatedly replace it with the sum 
of the squares of its digits. If the process loops endlessly and never reaches 1, it is called an unhappy (or sad) number."""
n=int(input("Enter Number: "))
num=n
flag=False
sum=0
while (True):
    while(n!=0):
        dig=n%10
        sum+=dig**2
        n//=10
    if sum==1:
        flag=True
        break
    elif sum==4:
        flag=False
        break
    else:
        n=sum
        sum=0
if(flag==True):
    print(num,"is a happy number")
else:
    print(num,"is not a happy number")
```
