## Working With Files
- Text Files: Text terminated with a special character '\n'
- Binary Files: Machine understandable binary language  
### File Access Modes
- Read Mode (r)
- Read and Write (r+)
- Write (w)
- Write and Read (w+) (whenever we open file in w+ mode, all the data is deleted and the data gets overridden meanwhile in R+ mode we can read and write)  
- Append (a) (append always adds data at the end of the file)  
- Append and Read (a+)

- f.seek(index): it moves cursor to given index
- f.read(index): reads the amount of index provided after the initial position of cursor
- f.tell(): it tells the position of cursor
- w+ mode will erase all data once the file is opened but after that we can do both read and write

- readline(index): Reads line if there is no index and reads only first line till count of index if any index given.
- readlines(index): Reads all lines and stores it in form of list, if index given then it reads all the lines within the count of index

```python
f=open('Text.txt')
x=f.readline(120)
print(x)
print(f.readline())
```

- write(str): writes in string format and goes to new line using \n
- writelines(list): writes in list format and doesm't to new line after using \n in the list

```python
details={"Name":"Alice","Age":21,"Degree":"Bachelor","Branch":"CSE","University":"Northern East University"}
f=open("copy.txt","w+")
for i,j in details.items():
    f.write(str(i)+": "+str(j)+"\n")
f.close()
```

```python
#WAP to copy the content of one file to another file but linewise reverse
f1=open("Text.txt")
f2=open("copy.txt","w") 
x=f1.readlines()
for i in x[::-1]:
    if(i.endswith("\n")):
        f2.write(i)
    else:
        f2.write(i+"\n")
f2.close()
```

```python
# replace python with java in Text.txt
f1=open("Text.txt","r") 
x=f1.readlines()
f1.close()
f2=open("Text.txt","w+") 
for i in x:
    print(i)
    f2.write(i.replace('Python','Java'))
f2.close()
```

```python
#WAP to count lines, words, characters and spaces from the txt file
f1=open("Text.txt")
x=f1.readlines()
wordc=0
linec=len(x)
spacec=0
charc=0
f1.seek(0)
for i in range(0,len(x)):
    a=f1.readline()
    x=a.split()
    wordc=wordc+len(x)
    spacec=spacec+a.count(" ")
    charc=charc+(len(a)-a.count(' ')-a.count('\n'))
print("Lines:",linec)
print("Words:",wordc)
print("Spaces:",spacec)
print("Characters:",charc)
```
