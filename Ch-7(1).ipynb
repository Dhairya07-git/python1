## Modules and Directories
- OS model in Python provides function for interacting with the operating system.
- OS comes under Python's standard utility modules.
### Handling the current working directory
- To get the location of the current working directory **os.getcwd()** is used
### Changing the current working directory
- To change the current working directory **os.chdir()** method is used.
- This method changes the cwd to a specific path.
- It only takes a single argument as a new directory path
### Creating the directory
- **os.mkdir()**:  
This method is used to create diretory named path with the specified folder name.  
This method raises FileExistsError if the directory to be created already exists.
### Listing our files and directories wiht python
- **os.listdir()**:  
This method used to get the list of all files and directories in the specified directory.  
If we don't specify any directory then the list of files and directories in the current working directory will be returned.
### Deleting directory or files
- **os.remove()**:  
This method is used to remove or delete a file. This method can't remove or delete a directory.  
If specified path is directory then OSError will be raised by the method.
- **os.rmdir()**:  
This method is used to remove or delete an empty directory OSError will be raised if the specified path is not an empty directory.

### Creating a user defined module
- The extension of file should be **".py"**.
- Whatever content that is written below %%writefile, it will be stored in txt file.
### Datetime Module
- date: Represents calender date.
- time: Represents time of day.
- datetime: combination of current date and time.

```python
%%writefile mymodule.py
def add(a,b):
    print(a+b)
def sub(a,b):
    print(a-b)
def mul(a,b):
    print(a*b)
def div(a,b):
    print(a/b)
def power(a,b):
    print(a**b)
def modulo(a,b):
    print(a%b)
```
