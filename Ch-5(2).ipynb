### Write a program to print the largest even and largest odd number from the list

```python
l=[12,45,67,89,123]
maxeve=0
maxodd=0
for i in l:
    if i%2==0:
        if i>maxeve:
            maxeve=i
    else:
        if i>maxodd:
            maxodd=i
print(maxeve)
print(maxodd)
```

### Write a program to perform a circular shift on a list to the right direction. Shift is to be entered by user

```python
n=int(input("Enter number of shifts:"))
a=[1,2,3,4,5,6,7]
n%=len(a)
a =a[-n:]+a[:-n]
print(a)
```

### Write a program to remove all occurences of an element from the given list

```python
n=int(input("Enter number that you want to remove:"))
l1=[1,2,5,3,1,2,5,3,7,7,2,1,8]
l2=l1.copy()
count=0
for i in l1:
    if i==n:
        count+=1
for i in range(count):
    l2.remove(n)
print(l2)
```

### Write a program to count number of string where the string length is 3 or more and the first and the last character are same from the given list of strings

```python
l=["xyz","abc","aba","2112","12345","1231"]
count=0
for i in l:
    if len(i)>=3:
        if i[0]==i[-1]:
            count+=1
print(count)
```
