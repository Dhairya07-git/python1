```python
#WAP to read a text file and reverse the words starting with i
f1=open("Demo.txt")
a=f1.readlines()
f1.close()

with open("Demo.txt","w+") as f2:
    for i in a:
        b=i.split()
        for j in range(len(b)):
            if b[j].startswith("i") or b[j].startswith("I"):
                w=b[j]
                b[j]=w[::-1]
        x=" ".join(b)+"\n"
        print(x)
        f2.write(x)
```

```python
#WAP to find most frequently occurring word from the given file
f=open("Text.txt")
words=f.read().lower().split()
freq={}
for i in words:
    i=i.strip(".,/?!()")
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
x=list(freq.values())
fmax=max(x)
for i in freq:
    if freq[i]==fmax:
        print("Most occurring word: \""+i+"\"")
```

```python
#WAP to store employee details in text file and calculate the average salary of employees
f=open("Text.txt","w")
n=int(input("Enter number of employees you want to add:"))
total=0
for i in range(n):
    print("Enter details of employee "+str(i+1)+":")
    name="Employee Name:"+input("Enter name of employee:")+"\n"
    sal=int(input("Enter salary of employee:"))
    salary="Employee Salary:"+str(sal)+"\n"
    f.write("Details of employee "+str(i+1)+":\n")
    f.write(name)
    f.write(salary)
    total+=sal
f.close()
print("Average salary:",total/n)
```

```python
f=open("mbox-short.txt")
l=[]
data={}
for line in f:
    if line.startswith("From"):
        x=line.split()
        l.append(x[1].strip())
for i in l:
    i=i.strip(".,/?!()")
    if i in data:
        data[i]+=1
    else:
        data[i]=1
a=list(data.values())
maxm=max(a)
for i,j in data.items():
    if j==maxm:
        print("Most emails sent by:",i)
```
