## Different type of functions
### 1) Function with no parameter and no return type  
  
def printline():  
    s=input("Enter Name:")  
    print(s)  
printline()  
  
### 2) Function with parameters and no return type  
  
def printline(s):  
    print s  
s=input("Enter Name:")  
printline(s)  
  
### 3) Function with no parameters and with return type  
  
def printline():  
    s=input("Enter Name:")  
    return s  
t=printline()  
print(t)  
  
### 4) Function with parameters and with return type  
  
def printline(s):  
    return s  
s=input("Enter Name:")  
t=printline(s)  
print(t)

## Types of arguments
### 1) Positional Arguments
### 2) Keyword Arguments
### 3) Default Arguments
### 4) Variable Length Arguments

```python
# Positional Argument
def sub(a,b):
    print(a-b)
sub(50,30)
sub(30,50)
```

```python
# Keyword Argument
def wish(name,msg):
    print("Hello",name,msg)
wish(msg="Good Morning",name="Dhairya")
wish("Dhairya",msg="Good Morning")
# wish(name="Dhairya","Good Morning") doesnt work
```

```python
# Default Argument
def wish1(name="Guest"):
    print("Hello",name,"Good Morning")
wish1("Yash")
wish1()
def wish2(name,msg="Good Morning"):
    print("Hello",name,msg)
wish2("Yash")
wish2("Yash","YOOOOOO")
```

```python
# Variable Length Argument
def fun(*n):
    total=0
    for n1 in n:
        total+=n1
    
```
