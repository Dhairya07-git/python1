### Accessing elements of list
1) using index
2) using slicing operation

```python
l=[10,20,30,40,50,60,70]
print(l[0])
print(l[-1])
print(l[-10])
```

```python
l=list(range(1,11))
print(l[1:4])
print(l[2:7:2])
print(l[8:2:-2])
print(l[4:100])
print(l[::-1])
print(l[-2:-8:-2])
```

```python
"""Even if we update in y it changes the list l because when we print id of l and y both of them are 
same also same thing is applicable for tuple but we just cant modify a tuple"""
l=[10,20,30,40]
y=l
y[0]=100
print(l)
print(id(l))
print(id(y))
```

## Important Functions for List
1) len() -> used to get length of the list
2) copy() -> to solve above problem this function is used to clone/duplicate
3) count() -> used for number of elements in list
4) index(element) -> gives index of first occurence of the element in list
5) append() -> used to add an element at the end of the list
6) insert(index,element) -> used to add an element at any place of the list (if index is out of bound on higher side it just appends at last and out of boumd on lower side then it adds at first position)
7) extend() -> used to combine two lists
8) remove(element) -> removes the first occurence of the element in the list
9) pop() -> removes last element from the list (gives error when used on empty list and it can also take index as parameter)
10) reverse() -> reverses entire list
11) clear() -> clears entire list
12) sort() -> sorts the data inside the list (sorted used for tuple sort used for list)  
we can also create function manually and use that function to sort
