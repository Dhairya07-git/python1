### Write a program to convert list of string to upper case using the map function

```python
s=["apple","banana","cherry","orange"]
u=list(map(lambda i:i.upper(),s))
print(u)
```

### Write a program to filter the words with length greater than 5

```python
s=["apple","banana","cherry","orange"]
def length(i):
    if len(i)>5:
        return i
s1=list(filter(length,s))
print(s1)
```

### Write a program to sort a list using lambda function

```python
n=[-6,1,-3,4,-2,-5]
a=lambda x:sorted(x)
print(a(n))
```

### Write a program to find number divisible by 19 or 13 from a list of numbers using lambda function

```python
n=[13,2,26,38,54,57,19,52,65,76,88]
a=lambda x:[num for num in x if num%19==0 or num%13==0] #List comprehension
print(a(n))
```

```python
n=[13,2,26,38,54,57,19,52,65,76,88]
a=list(filter(lambda x:(x%19==0 or x%13==0),n))
print(a)
```

### Write a program to sort a list of tuple by second element using lambda function

```python
l=[(1,3),(4,7),(8,5),(3,4),(8,1)]
s=sorted(l,key=lambda x:x[1])
print(s)
```
